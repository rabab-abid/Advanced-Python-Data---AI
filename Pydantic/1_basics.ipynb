```python
from pydantic import BaseModel
```

```python
class employee_schema(BaseModel):
    name: str
    id: int
    department: str
```

```python
obj= employee_schema(name="John", id=1, department="HR")
obj
```

## Coercion VS Strict Types

```python
obj2=employee_schema(**{"name": "John", "id": 1, "department": 101})
obj2
```

# Coercion

```python
obj3=employee_schema(**{"name": "John", "id": "1", "department": "HR"})
obj3
```

# Strict Type

```python
from pydantic import StrictInt
```

```python
class employee_schema(BaseModel):
    name: str
    id: StrictInt
    department: str
```

```python
obj4=employee_schema(name="John", id="1", department="HR")
```
